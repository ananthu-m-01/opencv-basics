```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray image",cmap="gray")
```

```python
sift = cv2.SIFT_create()
keypoints, descriptors = sift.detectAndCompute(gray_image, None)
sift_img = cv2.drawKeypoints(gray_image, keypoints, descriptors,None)
```

```python
show_image(sift_img,"SIFT features",cmap="gray")
```
