```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray image",cmap="gray")
```

```python
import numpy as np
def compute_glcm(image, offset=(1, 0), levels=256):
    glcm = np.zeros((levels, levels), dtype=np.float32)
    h, w = image.shape
    for i in range(h - offset[0]):
        for j in range(w - offset[1]):
            row, col = image[i, j], image[i + offset[0], j + offset[1]]
            glcm[row, col] += 1
    return glcm / glcm.sum()
```

```python
glcm = compute_glcm(gray_image)
```

```python
show_image(glcm,"GLCM image",cmap="gray")
```
