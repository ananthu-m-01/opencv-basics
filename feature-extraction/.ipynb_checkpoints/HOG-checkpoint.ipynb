```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.title(title)
    plt.imshow(img,cmap=cmap)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray Image",cmap="gray")
```

```python
from skimage.feature import hog
hog_features,hog_img = hog(gray_image,visualize=True )
```

```python
show_image(hog_img,"Histogram of Oriented Gradients",cmap="gray")
```
