```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray image",cmap="gray")
```

```python
from skimage.filters import gabor
gabor_image, _ = gabor(gray_image,frequency=0.05)
```

```python
show_image(gabor_image,"Gabor Filterd image",cmap="gray")
```
