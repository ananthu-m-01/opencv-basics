```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray image",cmap="gray")
```

```python
orb=cv2.ORB_create(nfeatures=1000)
keypoints,descriptors=orb.detectAndCompute(gray_image,None)
orb_img=cv2.drawKeypoints(gray_image,keypoints,descriptors,None)
```

```python
show_image(orb_img,"SURF",cmap="gray")
```
