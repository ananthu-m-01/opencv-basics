```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray image",cmap="gray")
```

```python
import numpy as np

def mean_filter(image, kernel_size=3):
    padded_image = np.pad(image, ((kernel_size//2,)*(2)), mode='constant')
    output = np.zeros_like(image)
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            roi = padded_image[i:i+kernel_size, j:j+kernel_size]
            output[i, j] = np.mean(roi)
    return output
```

```python
filtered_img = mean_filter(gray_image)
```

```python
show_image(filtered_img,"Filtered image",cmap="gray")
```
