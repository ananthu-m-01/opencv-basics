```python
import matplotlib.pyplot as plt
import cv2
import os
```

```python
image_path = os.path.join("..", "images", "leaf.webp")
image = cv2.imread(image_path)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
def show_image(img,title,cmap=None):
    plt.figure(figsize=(3,3))
    plt.imshow(img,cmap=cmap)
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
show_image(gray_image,"Gray image",cmap="gray")
```

```python
import numpy as np
dst = cv2.cornerHarris(np.float32(gray_image),blockSize=2,ksize=3,k=0.04)
dst = cv2.dilate(dst, None)
```

```python
hessian_like = image.copy()
```

```python
show_image(dst,"Haris and Hessian",cmap="gray")
```
